# Task 2 — Data Cleaning


## Cleaning Missing Values

```python
# Checking
df.isna().sum()
```

```python
# Handling nulls
df["age"]=df["age"].fillna(df["age"].median())
df["gender"]=df["gender"].fillna(df["gender"].mode()[0])
df["likes"]=df["likes"].fillna(df["likes"].median())
df["comments"]=df["comments"].fillna(df["comments"].median())
df["shares"]=df["shares"].fillna(df["shares"].median())
df["sentiment"]=df["sentiment"].fillna(df["sentiment"].mode()[0])
```

- All nulls very filled

## Data Formatting

```python
# Checking incorrect DataTypes
df.dtypes
```

```python
# Fixing incorrect datatypes
df["age"]=df["age"].astype("int64")
df["likes"]=df["likes"].astype("int64")
df["comments"]=df["comments"].astype("int64")
df["shares"]=df["shares"].astype("int64")
```

# Final Insights

```python
# 1. Content Performance

print("Post Type Performance")
print(df.groupby('post_type')[['engagement_score','engagement_rate']].mean())

print("\nCategory Performance")
print(df.groupby('post_category')[['engagement_score','engagement_rate']].mean())

print("\nCountry Performance")
print(df.groupby('country')[['engagement_rate']].mean())
```

## Content Performance

Post Type Performance
- Video posts have the highest average engagement score (12,664.58) and engagement rate (1.122).
- Reel posts have the lowest average engagement score (12,524.01) and engagement rate (0.78)
- Therefore, video posts perform best among the four post types.
  
Category Performance
- Music has the highest average engagement score (12,788.54).
- Food has the highest average engagement rate (1.359).
- Travel has the lowest average engagement rate (0.702).
- Therefore, Music performs best in engagement score, while Food performs best in engagement rate.
  
Country Performance
- Brazil has the highest average engagement rate (1.541).
- Australia ranks second (1.324).
- USA has the lowest average engagement rate (0.577).
- Therefore, Brazil shows the strongest average engagement rate among the countries.

```python
# 2. User Trends
print(df.groupby('age')['engagement_rate'].mean().round(4))
print(df.groupby('is_verified')[['engagement_score','engagement_rate']].mean().round(4))
```

## User Trends

Age and Engagement
- Users of age 24 years have the highest average engagement rate (2.8973).
- Age 64 has the lowest average engagement rate (0.4025)..
- This indicates that younger users, particularly those who are of age 24, show higher average engagement scores.

Verified vs Non-Verified Accounts
- Verified accounts have a higher average engagement rate (1.054) than non-verified accounts (0.955).
- Non-verified accounts have a higher average engagement score (12,644.30) than verified accounts (12,309.27).
- Therefore, verified accounts perform better in engagement rate, while non-verified accounts perform better in engagement score.

```python
# 3. Behavioral Insights

df['hour'] = df['posted_at'].dt.hour

print("\nImpressions by Hour")
print(df.groupby('hour')['impression_count'].mean())

print("\nWatch Time by Device")
print(df.groupby('device_type')['watch_time_sec'].mean())
```

## Behavioral Insights

Best Time of Day for Impressions
- The dataset shows an average of 50,013.73 impressions for hour 0.
- However, all records have hour 0 because the dataset contains dates without actual posting times.
- Therefore, the best time of day for impressions cannot be identified.

Device Type Impact on Watch Time
- Mobile has the highest average watch time (4,087.83 seconds).
- Tablet comes second (3,979.74 seconds).
- Desktop has the lowest average watch time (3,974.79 seconds).
- This indicates that users spend more time watching posts on mobile devices.

```python
# 4. Sentiment Analysis


print(df.groupby('sentiment')[[ 'engagement_score','engagement_rate']].mean())

print(df[df['sentiment'].isin(['negative', 'neutral'])]
      .groupby('sentiment')[['likes', 'comments', 'shares', 'watch_time_sec']]
      .mean().round(2))
```

## Sentiment Analysis

Average Engagement by Sentiment
- Negative sentiment has the highest average engagement score (12,731.48) and engagement rate (1.038).
- Neutral sentiment has the lowest average engagement score (12,448.15).
- Positive sentiment has the lowest average engagement rate (0.920).

Negative and Neutral Posts
- Negative posts receive more average likes (10,208.10) and shares (1,007.31) compared to neutral posts (9,923.19 likes and 996.57 shares).
- Neutral posts receive slightly more average comments (1,528.40) than negative posts (1,516.09).
- Negative posts have slightly higher average watch time (4,054.64 seconds) compared to neutral posts (4,031.82 seconds), indicating marginally better viewing engagement.

# Overall Conclusion

The analysis of social media engagement data helped identify patterns in content performance, user behaviour, device usage and audience sentiment. Video posts achieved the highest average engagement rate, while Food was the best-performing category in terms of engagement rate. Brazil recorded the highest average engagement rate among the countries analysed.

Age-wise engagement varied, with 24-year-old users showing the highest average engagement rate. Verified accounts performed better in engagement rate, whereas non-verified accounts had a higher average engagement score. Mobile users recorded the highest average watch time. However, the best time of day for impressions could not be identified because the dataset contained dates without actual posting times.

In sentiment analysis, negative posts showed the highest average engagement score and engagement rate, followed by neutral posts. Negative posts also received slightly more likes, shares and watch time than neutral posts, while neutral posts received slightly more comments.

Overall, the analysis shows that engagement performance varies across post types, categories, countries, age groups, devices and sentiment. These findings can help understand audience behaviour and identify content patterns that receive better engagement.
